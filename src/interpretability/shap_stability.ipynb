```python
import pickle
import numpy as np
```

# Get the data

```python
exp = "DIMAFx"
data = 'brca'
```

```python
print("Unimodal")
results = {}
for k in [5, 10, 20, 30]:
    means = []
    results = f"& $k  = {k}$ "
    for i in range(5):
        shap_results_fold_dir = f'../results/dss_survival_{data}/{exp}/Fold_{i}/post_training/shap/modal/shap_stability_modal.pkl'
        shap_stab_dict = pickle.load(open(shap_results_fold_dir, 'rb'))     
        results += f"& {shap_stab_dict[f'Top_{k}']['mean']:.3f}±{shap_stab_dict[f'Top_{k}']['std']:.3f} "
        means.append(shap_stab_dict[f'Top_{k}']['mean'])

    mean_over_folds = np.mean(means)
    std_over_folds = np.std(means)
    results += f"& {mean_over_folds:.3f}±{std_over_folds:.3f} \\\\"

    
    print(results)
```

```python
print("Multimodal")
results = {}

for k in [5, 10, 20, 30]:
    means = []
    results = f"& $k  = {k}$ "
    for i in range(5):
        shap_results_fold_dir = f'../results/dss_survival_{data}/{exp}/Fold_{i}/post_training/shap/post_attn/shap_stability_post_attn.pkl'
        shap_stab_dict = pickle.load(open(shap_results_fold_dir, 'rb'))     
        results += f"& {shap_stab_dict[f'Top_{k}']['mean']:.3f}±{shap_stab_dict[f'Top_{k}']['std']:.3f} "
        means.append(shap_stab_dict[f'Top_{k}']['mean'])

    mean_over_folds = np.mean(means)
    std_over_folds = np.std(means)
    results += f"& {mean_over_folds:.3f}±{std_over_folds:.3f} \\\\"

    
    print(results)
```
